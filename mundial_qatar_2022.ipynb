### Tercera Parte

```python
# Contar cuantos goles hizo cada equipo en total en el mundial
equipos = pd.concat([mundial['1'], mundial['2']]).unique()
goles_equipo = {}

for equipo in equipos:
    goles_local = mundial[mundial['1'] == equipo]['1_goals'].sum()
    goles_visitante = mundial[mundial['2'] == equipo]['2_goals'].sum()
    goles_equipo[equipo] = goles_local + goles_visitante

goles = pd.Series(goles_equipo)
goles
```

```python
# Graficar los goles por equipo en todo el mundial
fig, ax = plt.subplots()
bar_labels = goles.index
counts = goles.values
ax.bar(bar_labels, counts)
ax.set_xlabel("Selección")
ax.set_ylabel("Goles")
plt.xticks(rotation=90)
plt.show()
```

```python
# Cual es el equipo que hizo mas goles en el mundial?
goles.idxmax()
```

```python
# Por cada partido, cual fue la diferencia de goles?
diferencia_goles = mundial['1_goals'] - mundial['2_goals']
abs(diferencia_goles)
```

```python
# Por cada partido, cual fue el equipo que gano?
for i in range(len(mundial)):
    if mundial.loc[i, "1_goals"] > mundial.loc[i, "2_goals"]:
        ganador = mundial.loc[i, "1"]
    elif mundial.loc[i, "2_goals"] > mundial.loc[i, "1_goals"]:
        ganador = mundial.loc[i, "2"]
    else:
        ganador = "Empate"

    print("Partido", mundial.loc[i, "match_no"], ":", ganador)
```

```python
# En cada fase del mundial, cual fue el equipo que hizo mas goles?
mundial["group"] = mundial["group"].apply(
    lambda x: "group" if "Group" in x else x
)

fases = mundial["group"].unique()

for fase in fases:
    partidos = mundial[mundial["group"] == fase]

    goles = {}

    for i in range(len(partidos)):
        equipo1 = partidos.iloc[i]["1"]
        equipo2 = partidos.iloc[i]["2"]

        goles[equipo1] = goles.get(equipo1, 0) + partidos.iloc[i]["1_goals"]
        goles[equipo2] = goles.get(equipo2, 0) + partidos.iloc[i]["2_goals"]

    ganador = max(goles, key=goles.get)

    print(fase,":", ganador,":", goles[ganador], "goles")
```
